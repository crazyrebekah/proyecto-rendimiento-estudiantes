```python
# ============================================================
# PREPROCESAMIENTO DEL DATASET DE ESTUDIANTES
# ============================================================


# 1. IMPORTACIÓN DE LIBRERÍAS

import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files


# 2. CONFIGURACIÓN

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")


# 3. SUBIR Y CARGAR EL DATASET

print("Selecciona el archivo dataset_estudiantes.csv")

archivos_subidos = files.upload()

nombre_archivo = list(archivos_subidos.keys())[0]

df = pd.read_csv(
    io.BytesIO(archivos_subidos[nombre_archivo])
)

print("\nDataset cargado correctamente.")

display(df.head())

print(
    f"\nDimensiones iniciales: "
    f"{df.shape[0]} filas y {df.shape[1]} columnas."
)


# 4. CREAR UNA COPIA DE SEGURIDAD

df_limpio = df.copy()


# 5. ELIMINAR FILAS DUPLICADAS

duplicados_antes = df_limpio.duplicated().sum()

df_limpio = df_limpio.drop_duplicates().copy()

print(
    f"\nFilas duplicadas eliminadas: "
    f"{duplicados_antes}"
)


# 6. COMPROBAR VALORES NULOS ANTES DEL TRATAMIENTO

print("\nValores nulos antes del preprocesamiento:")

display(
    df_limpio.isnull()
    .sum()
    .to_frame("cantidad_nulos")
)


# 7. IMPUTACIÓN DE HORAS DE SUEÑO

# Es una variable numérica, por lo que utilizamos la mediana.
# La mediana es menos sensible a los valores extremos.

mediana_horas_sueno = df_limpio["horas_sueno"].median()

df_limpio["horas_sueno"] = (
    df_limpio["horas_sueno"]
    .fillna(mediana_horas_sueno)
)

print(
    "\nMediana utilizada en horas_sueno:",
    round(mediana_horas_sueno, 2)
)


# 8. IMPUTACIÓN DE VARIABLES CATEGÓRICAS

# Se utiliza la categoría "Desconocido" para conservar la información
# de que el valor original no estaba disponible.

columnas_categoricas_con_nulos = [
    "horario_estudio_preferido",
    "estilo_aprendizaje"
]

for columna in columnas_categoricas_con_nulos:

    df_limpio[columna] = (
        df_limpio[columna]
        .fillna("Desconocido")
    )


# 9. COMPROBAR QUE YA NO QUEDAN NULOS

print("\nValores nulos después del preprocesamiento:")

display(
    df_limpio.isnull()
    .sum()
    .to_frame("cantidad_nulos")
)

print(
    "\nNúmero total de valores nulos:",
    df_limpio.isnull().sum().sum()
)


# 10. COMPROBAR LOS TIPOS DE DATOS

print("\nTipos de datos:")

display(
    df_limpio.dtypes
    .to_frame("tipo_de_dato")
)


# 11. COMPROBAR VALORES DE LAS VARIABLES CATEGÓRICAS

columnas_categoricas = (
    df_limpio
    .select_dtypes(include="object")
    .columns
    .tolist()
)

for columna in columnas_categoricas:

    print(f"\nValores de {columna}:")

    display(
        df_limpio[columna]
        .value_counts(dropna=False)
        .to_frame("cantidad")
    )


# 12. COMPROBAR RANGOS DE LAS VARIABLES NUMÉRICAS

columnas_numericas = (
    df_limpio
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

rangos = pd.DataFrame({
    "mínimo": df_limpio[columnas_numericas].min(),
    "máximo": df_limpio[columnas_numericas].max()
})

print("\nRangos de las variables numéricas:")

display(rangos)


# 13. CREAR DATASET PARA REGRESIÓN

# El objetivo de regresión es nota_final.
# Eliminamos aprobado para evitar fuga de información.

df_regresion = df_limpio.drop(
    columns=["aprobado"]
).copy()

print("\nDataset preparado para regresión:")

print(
    f"{df_regresion.shape[0]} filas y "
    f"{df_regresion.shape[1]} columnas."
)

display(df_regresion.head())


# 14. CREAR DATASET PARA CLASIFICACIÓN

# El objetivo de clasificación es aprobado.
# Eliminamos nota_final porque determina directamente el aprobado.

df_clasificacion = df_limpio.drop(
    columns=["nota_final"]
).copy()

print("\nDataset preparado para clasificación:")

print(
    f"{df_clasificacion.shape[0]} filas y "
    f"{df_clasificacion.shape[1]} columnas."
)

display(df_clasificacion.head())


# 15. GUARDAR LOS DATASETS PREPROCESADOS

df_regresion.to_csv(
    "df_regresion_estudiantes.csv",
    index=False
)

df_clasificacion.to_csv(
    "df_clasificacion_estudiantes.csv",
    index=False
)

print("\nArchivos creados correctamente:")

print("- df_regresion_estudiantes.csv")
print("- df_clasificacion_estudiantes.csv")


# 16. DESCARGAR LOS ARCHIVOS

files.download("df_regresion_estudiantes.csv")
files.download("df_clasificacion_estudiantes.csv")
```

```python
## Conclusiones del preprocesamiento

No se encontraron filas duplicadas. Los valores ausentes de `horas_sueno` se sustituyeron por la mediana, al tratarse de una variable numérica. En las variables categóricas `horario_estudio_preferido` y `estilo_aprendizaje`, los valores ausentes se sustituyeron por la categoría `Desconocido`.

Se crearon dos conjuntos de datos diferentes. Para regresión se eliminó `aprobado`, evitando utilizar una variable derivada de la nota final. Para clasificación se eliminó `nota_final`, ya que esta variable determina directamente si el estudiante está aprobado.

La codificación de las variables categóricas y el escalado se realizarán dentro de los pipelines de modelado. De esta manera, dichos transformadores se ajustarán únicamente con los datos de entrenamiento y se evitará la fuga de información.
```
